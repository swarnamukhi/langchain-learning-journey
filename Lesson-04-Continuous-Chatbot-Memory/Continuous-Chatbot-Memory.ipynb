```python
import vertexai
```

```python
from langchain_google_vertexai import ChatVertexAI
from langchain_core.runnables.history import RunnableWithMessageHistory
from langchain_core.messages import HumanMessage, SystemMessage
from langchain_core.chat_history import InMemoryChatMessageHistory
```

```python
model=ChatVertexAI(model="gemini-2.5-flash")
```

```python
system_instruction=""" You are a helpful, knowledgeable, and patient LangChain tutor who teaches users how to build Generative AI applications using LangChain with Gemini models.

Task:
Help the user build LangChain applications using Gemini models remember user using models from vertexai platform.

Please follow these instructions:

1. Provide complete code based on the user's request.
2. Explain the code line by line.
3. Explain all packages and imports used.
4. Explain the flow of execution step by step.
5. Keep explanations beginner-friendly and clear.
6. Use simple language whenever possible.
7.Provide real-world examples whenever possible.
8. Explain concepts before jumping into advanced code.
  """
```

```python
store={}
def get_session_history(session_id):
    if session_id not in store:
        store[session_id]=InMemoryChatMessageHistory()
    return store[session_id]
```

```python
chain=RunnableWithMessageHistory(model,get_session_history)
```

```python
from IPython.display import display,Markdown
```

```python
session_id = 1209

while True:


  user_input = input("You: ")

  if user_input.lower() == "exit":
      print("Chat ended.")
      break

  response = chain.invoke(
      [
          SystemMessage(content=system_instruction),
          HumanMessage(content=user_input)
      ],
      config={
          "configurable": {
              "session_id": session_id
          }
      }
  )

  display(Markdown(response.content))
  print("\n" + "="*80)
  print("Type your next question below ↓")
  print("="*80 + "\n")
```
