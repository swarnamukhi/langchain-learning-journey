```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import HumanMessage, SystemMessage
from IPython.display import Markdown, display
```

#langchain message classes with differnt models from diffrent platforms
1.model from GoogleAiStudio

```python
llm=ChatGoogleGenerativeAI(api_key="api key"
,model="gemini-3-flash-preview")
```

```python
learning=[SystemMessage(content="you are a GenAI tutor"),
          HumanMessage(content="what are the important topics to learn in Genai") ]
```

```python
content_creation=[SystemMessage(content="you are a content generator to help the content creators to write blogs"),
                  HumanMessage(content="write a blog on leh ladhak beauty") ]
```

```python
tutor_model=llm.bind(temperature=0,max_output_tokens=5000)
```

```python
content_model=llm.bind(temperature=0.3,max_output_tokens=5000)
```

```python
tutor_response= tutor_model.invoke(learning)
```

```python
display(Markdown(tutor_response.text))
```

```python
content_model_response=content_model.invoke(content_creation)
```

```python
display(Markdown(content_model_response.text))
```
